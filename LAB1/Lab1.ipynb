### Task #2: Loading and Visualizing Images
#### a. Loading/reading an image using OpenCV

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('images/cameraman.tif')  # Load the image data into a NumPy array

plt.imshow(img)
plt.show()
```

**Note on colors:** OpenCV reads images in **BGR** order, but `matplotlib` expects **RGB**. Since `cameraman.tif` is grayscale, OpenCV still loads it as a 3-channel image, so the colors above may look tinted. Convert it for a correct grayscale display:

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(img_rgb, cmap='gray')
plt.title('Cameraman (OpenCV, corrected)')
plt.show()
```

#### b. Loading/reading image using PIL

```python
from PIL import Image
import matplotlib.pyplot as plt
import matplotlib.cm as cm

img2 = Image.open('images/lena_gray_256.tif')  # load image into a PIL Image object

plt.imshow(img2, cmap=cm.Greys_r)
plt.show()
```
